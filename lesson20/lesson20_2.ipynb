```python
import numpy as np
import pandas as pd

df = pd.DataFrame({'key1':['a', 'a', 'b', 'b', 'a'],
                  'key2':['one', 'two', 'one', 'two', 'one'],
                  'data1': np.random.randn(5),
                  'data2':np.random.randn(5)})
df.groupby('key1')[['data1', 'data2']].sum()
df1 = df.groupby(['key1', 'key2']).sum()
df1.columns.name = 'data'
df1.stack().unstack(level=['key1', 'data'])

df1.groupby(['key1', 'key2']).agg([('加總', 'sum'), ('平均', 'mean')]) #.agg可以放多個統計
```
