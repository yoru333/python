```python
import numpy as np
import pandas as pd
def max_min(value):
    return value.max() - value.min()

df = pd.read_csv('./data/tips.csv')
df.columns = ['總票價', '小費', '吸煙者', '日期', '時間', '大小']
grouped = df.groupby(['日期', '吸煙者'])
df1 = grouped['小費'].agg([('平均', 'mean'), ('加總', 'sum'), ('最大減最小', max_min)])
df1.unstack(level='日期')
functions = [('數量','count'),('平均','mean'),('最大值','max')]
df2 = grouped[['小費','總票價']].agg(functions)
```
