```python
#Styler.format()
#Style.format_index()
#Style.relabel_index()

import pandas as pd
import numpy as np 

df = pd.DataFrame({
    "strings": ["Adam", "Mike"],
    "ints": [1, 3],
    "floats": [1.123, 1000.23]
})
type(df.style)

display(df)
s = df.style\
.format(precision=3, thousands=",", decimal='.')\
.format_index(str.upper,axis=1)\
.relabel_index(["列1","列2"])

s
```

```python

import pandas as pd
weather_df = pd.DataFrame(np.random.rand(10,2)*5,
                          index=pd.date_range(start="2024-01-01", periods=10),
                          columns=["Tokyo","Taipei"]
                          )
def rain_condition(v):
    if v < 1.75:
        return "乾燥"
    elif v < 2.75:
        return "小雨"
    else:
        return "大雨"

def make_pretty(styler):
    styler.set_caption("天氣狀況")
    styler.format(rain_condition) #「把 DataFrame 的值轉換為字串顯示，但不改變實際數值。
    return styler

s = weather_df.loc['2024-01-01':'2024-01-07'].style.pipe(make_pretty)
s
```
