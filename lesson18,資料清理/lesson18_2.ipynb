```python
import pandas as pd

df = pd.read_excel('各鄉鎮市區人口密度.xlsx',index_col=0) #	index_col=0 以第 0 欄為索引
df1 = df[["年底人口數","土地面積","人口密度"]]

def statistic(s):
    max_city = s[s == s.max()].index[0] #s[s == s.max()] 過濾資料 []裡面放篩選條件
    min_city = s[s == s.min()].index[0]
    return pd.Series([s.mean(),s.max(),max_city ,s.min(),min_city ,s.std(),s.median()],index=['平均','最高','最高縣市','最低','最低縣市','標準差','中間數'])

df2 = df1.apply(statistic,axis=0)
df2
```
