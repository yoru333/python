```python
#絕對位置
import pandas as pd
import os
current_dir = os.path.dirname(os.path.abspath(__name__))
#_name__ 在主執行檔通常是 "__main__"
#os.path.abspath(__name__) 會取得 "__main__" 的絕對路徑。
data_dir = os.path.join(current_dir,'data')
#在 current_dir 底下接一個資料夾名稱 'data'，組成一個新的完整路徑。
csv_filename = '台鐵車站資訊.csv'
csv_path = os.path.join(data_dir,csv_filename)
#連接csv_filename形成最終完整路徑
station_info = pd.read_csv(csv_path)
```

```python
import pandas as pd
csv1_filename = '每日各站進出站人數2020.csv'
csv1_path = os.path.join(data_dir, csv1_filename)
inOut2020_df = pd.read_csv(csv1_path)

merge2022 = pd.merge(inOut2020_df, station_info, how="left", left_on="staCode",right_on="stationCode")
#left_on="staCode" 左表要對應合併的欄位名稱
#right_on="stationCode" 右表要對應合併的欄位名稱
df1 = merge2022.reindex(columns=['trnOpDate','gateInComingCnt','gateOutGoingCnt','stationName'])
df1.columns = ['日期','進站人數','出站人數','站名']
df1['日期'] = pd.to_datetime(df1['日期'].astype(str))
df2 = df1.set_index('日期')
df3 = df2.groupby('站名').sum() #同樣名稱的站名做加總
df4 = df2[['進站人數','出站人數']].resample('ME').sum()
#.resample 把「原本依照時間記錄的一筆筆資料」，重新按照比較大的時間單位來分組
#'ME' 指Month End(每月月底)
#'MS' = Month Start（每月月初）

df2['年月'] = df2.index.to_period('M') #.to_period 轉成月份-->2025-01
df5 = df2.groupby('年月')[['進站人數','出站人數']].sum()
#多個欄位 [['進站人數','出站人數']]，就要雙層中括號
df5
```
